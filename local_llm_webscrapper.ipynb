```python
# Website Summarizer using Generative AI (Local LLM)

## Overview
This notebook demonstrates a simple Generative AI application that summarizes the main content of a given website URL using a **locally hosted Large Language Model (LLM)** via Ollama.

The goal of this project is to explore how LLMs can be applied to real-world text summarization tasks while understanding practical challenges such as context limits and noisy webpage content.

---

## Problem Statement
Webpages often contain long-form content mixed with navigation menus, ads, and irrelevant text.
Manually reading and extracting key information can be time-consuming.

This project aims to automatically:
- Fetch webpage content from a URL
- Extract meaningful text
- Generate a concise, human-readable summary using a Generative AI model

---

## Approach
1. Fetch raw HTML content from the given website URL
2. Extract and clean meaningful textual content
3. Send the cleaned text to a local LLM using a structured prompt
4. Generate a concise summary of the webpage

---

## Why a Local LLM?
This project uses a locally hosted LLM (via Ollama) to:
- Avoid dependency on paid cloud APIs
- Enable offline experimentation
- Gain hands-on understanding of local LLM workflows

---

## Technologies Used
- Python
- Large Language Models (LLMs)
- Ollama (Local LLM)
- Prompt Engineering
- Tokenization & Context Handling

---

## Example Usage
Input:
- Website URL (e.g., a blog, documentation page, or article)

Output:
- A short, readable summary capturing the main idea of the webpage

---

## Limitations
- Summarization quality depends on webpage structure
- Very large webpages may exceed model context limits
- Single-page summarization only (no multi-page crawling)

---

## Future Improvements
- Add text chunking for large webpages
- Support multiple URLs
- Convert into a simple web or CLI application
- Improve content extraction accuracy
```

```python
import os
from dotenv import load_dotenv
from scraper import fetch_website_contents
from IPython.display import Markdown, display
from openai import OpenAI
```

```python
OLLAMA_BASE_URL = "http://localhost:11434/v1"

ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='ollama')
```

```python
system_prompt = """
You are an advisor assistant that analyzes the contents of a website,
and provides a short, snarky, humorous summary, ignoring text that might be navigation related.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""


user_prompt_prefix = """
Here are the contents of a website.
Provide a short summary of this website.
If it includes news or announcements, then summarize these too.
A comment from your end
"""


web = fetch_website_contents("https://artefacts.data.gov.in/state-ut-wise-cyber-crimes-in-india-from-2016-to-2018/")
```

```python

def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]

messages_for(web)
```

```python
def summarize(url):
    website = fetch_website_contents(url)
    response = ollama.chat.completions.create(
        model = "llama3.2",
        messages = messages_for(website)
    )
    return response.choices[0].message.content
```

```python
def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
```

```python
display_summary("https://artefacts.data.gov.in/state-ut-wise-cyber-crimes-in-india-from-2016-to-2018/")
```
