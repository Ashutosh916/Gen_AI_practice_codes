```python
This notebook explores the problem-solving and reasoning capabilities  
of **Large Language Models (LLMs)** using both a **local LLM (Ollama)**  
and a **cloud-based LLM (Gemini API)**.

Logic puzzles and structured questions are used to evaluate how  
different models respond, reason, and explain their solutions.

The project integrates the respective model frameworks and API keys  
to enable programmatic interaction and comparison.
```

```python
import os
import requests
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
load_dotenv()

gemini_api_key = os.getenv('GEMINI_API_KEY')

OLLAMA_BASE_URL = "http://localhost:11434/v1"

ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='ollama')

if gemini_api_key:
    print(f"Gemini API Key exists")
else:
    print("Gemini API Key not set (and this is optional)")
```

```python
gemini_url = "https://generativelanguage.googleapis.com/v1beta/openai/"

gemini = OpenAI(api_key=gemini_api_key, base_url=gemini_url)
```

```python
question_prompt = """
Three people check into a hotel room that costs ₹300.
They each pay ₹100.
Later, the hotel realizes the room should have cost ₹250 and gives ₹50 to the bellboy to return.
The bellboy keeps ₹20 and gives ₹10 back to each person.

Each person has now paid ₹90, so total paid is ₹270.
Add the ₹20 kept by the bellboy = ₹290.
Where did the remaining ₹10 go?
"""

question = [
    {"role": "user", "content": question_prompt},
]
```

```python
response = ollama.chat.completions.create(model="llama3.2", messages=question)
display(Markdown(response.choices[0].message.content))
```

```python
response = gemini.chat.completions.create(model="gemini-2.5-flash", messages=question)
display(Markdown(response.choices[0].message.content))
```
